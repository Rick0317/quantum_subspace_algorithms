```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

from DATA_ENERGIES import h2o_bond_lengths, n2_bond_lengths
from DATA_MEASUREMENTS import *
```

```python
start = 0
stop  = 17

plot_parameters = {'font.family': 'stixgeneral',
                   'font.serif': 'cmr10',
                   'mathtext.fontset': 'cm',
                   }
plt.rcParams.update(plot_parameters)

fig, ax = plt.subplots(figsize=(8,6))

ax.plot(h2o_bond_lengths[start:stop], H2O_VO_COSTS[start:stop], marker='o', markersize=8, linewidth=2, color='dodgerblue')
ax.plot(h2o_bond_lengths[start:stop], H2O_PT_COSTS[start:stop], marker='s', markersize=8, linewidth=2, color='red')

plt.ylabel(r'$\epsilon^2_{\text{Q}}M$', fontsize=24)
# plt.xlabel(r'Bond Length ($\AA$)', fontsize=24)
ax.tick_params(axis='both', which='major', labelsize=20)

plt.savefig('output_h2o_measurement_cost.pdf', dpi=400, bbox_inches='tight')
```

```python
start = 0
stop  = 15

plot_parameters = {'font.family': 'stixgeneral',
                   'font.serif': 'cmr10',
                   'mathtext.fontset': 'cm',
                   }
plt.rcParams.update(plot_parameters)

fig, ax = plt.subplots(figsize=(8,6))

ax.plot(n2_bond_lengths[start:stop], N2_VO_COSTS[start:stop], marker='o', markersize=8, linewidth=2, color='dodgerblue')
ax.plot(n2_bond_lengths[start:stop], N2_PT_COSTS[start:stop], marker='s', markersize=8, linewidth=2, color='red')

plt.ylabel(r'$\epsilon^2_{\text{Q}}M$', fontsize=24)
plt.xlabel(r'Bond Length ($\AA$)', fontsize=24)
ax.tick_params(axis='both', which='major', labelsize=20)

plt.savefig('output_n2_measurement_cost.pdf', dpi=400, bbox_inches='tight')
```

```python
avo_style = {'color': 'dodgerblue', 'marker': 'o', 'markersize': 8, 'linewidth': 2, 'linestyle': '-'}
pt_style = {'color': 'red', 'marker': 's', 'markersize': 8, 'linewidth': 2, 'linestyle': '-'}

line_avo = plt.Line2D([0], [0], **avo_style, label='VO')
line_pt = plt.Line2D([0], [0], **pt_style, label='PT')

fig_legend = plt.figure(figsize=(5, 0.5))

leg=fig_legend.legend(handles=[line_avo, line_pt], 
                  loc='center', 
                  ncol=2, 
                  frameon=False,
                  fontsize=25,
                  columnspacing=4)

plt.gca().set_axis_off()

plt.savefig('output_legend_sampling_cost.pdf', dpi=400, transparent=True)
```
