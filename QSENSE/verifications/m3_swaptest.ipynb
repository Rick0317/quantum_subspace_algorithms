Here, I test if, for real quantum states, if computing the matrix element, and computing the swap test expectation value via `XorY_augment` and `XandY_augment` always produce the same results.

```python
import sys
sys.path.append('../')

from utils_m3_swap import (
    XandY_augment,
    y_parity_of_term,
    XorY_augment
)
from utils_states import (
    create_composite_state,
    convert_dense_format_to_sparse_format
)
from openfermion import (
    get_sparse_operator,
    QubitOperator
)
from utils_basic import random_pauli_hamiltonian
import numpy as np
from numpy.random import uniform
```

```python
for _ in range(10):
    Nqubits  = 12
    Nterms   = 500

    psi      = convert_dense_format_to_sparse_format(uniform(-1, 1, 2 ** Nqubits))
    phi      = convert_dense_format_to_sparse_format(uniform(-1, 1, 2 ** Nqubits))
    comp     = create_composite_state(psi, phi, Nqubits)

    H        = random_pauli_hamiltonian(Nqubits, Nterms)
    H_AND    = XandY_augment(H, Nqubits)
    H_OR     = XorY_augment(H, Nqubits)

    print(len(H.terms), len(H_AND.terms), len(H_OR.terms))

    H_sp     = get_sparse_operator(H, Nqubits)
    H_AND_sp = get_sparse_operator(H_AND, Nqubits + 1)
    H_OR_sp  = get_sparse_operator(H_OR, Nqubits + 1)

    element     = (psi @ H_sp @ phi.T)[0,0]
    element_AND = (comp @ H_AND_sp @ comp.T)[0,0]
    element_OR  = (comp @ H_OR_sp @ comp.T)[0,0]

    assert np.allclose(element, element_AND)
    assert np.allclose(element, element_OR)
    assert np.allclose(element_AND, element_OR)
    assert not np.allclose(element, np.real(element))
    assert not np.allclose(element, 1j * np.imag(element))
```

```python
for _ in range(10):
    Nqubits  = 12
    Nterms   = 500

    psi      = convert_dense_format_to_sparse_format(uniform(-1, 1, 2 ** Nqubits))
    phi      = convert_dense_format_to_sparse_format(uniform(-1, 1, 2 ** Nqubits))
    comp     = create_composite_state(psi, phi, Nqubits)

    Hbase    = random_pauli_hamiltonian(Nqubits, Nterms)

    H = QubitOperator()
    for term, _ in Hbase.terms.items():
        H += (uniform(-1,1) + 1j * uniform(-1,1)) * QubitOperator(term)

    H_AND    = XandY_augment(H, Nqubits)
    H_OR     = XorY_augment(H, Nqubits)

    print(len(H.terms), len(H_AND.terms), len(H_OR.terms))

    H_sp     = get_sparse_operator(H, Nqubits)
    H_AND_sp = get_sparse_operator(H_AND, Nqubits + 1)
    H_OR_sp  = get_sparse_operator(H_OR, Nqubits + 1)

    element     = (psi @ H_sp @ phi.T)[0,0]
    element_AND = (comp @ H_AND_sp @ comp.T)[0,0]
    element_OR  = (comp @ H_OR_sp @ comp.T)[0,0]

    assert np.allclose(element, element_AND)
    assert np.allclose(element, element_OR)
    assert np.allclose(element_AND, element_OR)
    assert not np.allclose(element, np.real(element))
    assert not np.allclose(element, 1j * np.imag(element))
```
